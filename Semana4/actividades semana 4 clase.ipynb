```python
Opcion = 0

def suma(num1, num2):
    return num1 + num2

def resta(num1, num2):
    return num1 - num2

def multiplicacion(num1, num2):
    return num1 * num2

print ("=== MENU PRINCIPAL ===")
print ("1. SUmar dos numeros")
print ("2. Restar dos numeros")
print ("3. Multiplicar dos numeros")
print ("4. Salir")

opcion = int(input("Selecciona una opción: "))
num1 = float(input("Ingresa el primer número: "))
num2 = float(input("Ingresa el segundo número: "))

match opcion:
    case 1:
        print(f"El resultado de la suma es: {suma(num1, num2)}")
    case 2:
        print(f"El resultado de la resta es: {resta(num1, num2)}")
    case 3:
        print(f"El resultado de la multiplicación es: {multiplicacion(num1, num2)}")
    case 4:
        print("Saliendo del programa...")
```
