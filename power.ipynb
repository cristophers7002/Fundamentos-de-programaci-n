## CALCULADORA TIEMPO DIGITAL

```python
nombre = input( "Hola, cual es tu nombre?")
Redes = float(input("Cuantas horas le dedicas a las redes sociales al dia?"))
videojuegos = float(input("cuantas horas le dedicas a los viedeojuegos al dia?"))
escuela = float(input("cuantas horas le dedicas a la escuela al dia?"))
trabajo = float(input("cuantas horas le dedicas al trabajo al dia?"))
peliculas = float(input("cuantas horas le dedicas a ver peluculas/series al dia?"))

Horas = Redes + videojuegos + escuela + trabajo + peliculas
porcentajes = (Horas / 24) * 100

print("Hola", nombre, "ya vi que le dedicas", Horas, "horas a tus actividades diarias y eso es un", f"{porcentajes:.2f}", "% de tu dia")
```

## Cuenta total + propina 


```python
Total_de_cuenta = float(input("Cuanto es el total de tu cuenta?"))
propina = float(input("Cuanto le quieres dar de propina?"))
personas = float(input("Cuantas personas van a pagar la cuenta?")) 

total_de_propina = (Total_de_cuenta * propina) / 100
total_a_pagar = Total_de_cuenta + total_de_propina
pago_por_persona = total_a_pagar / personas

print ("el total de cuenta es de", f"{Total_de_cuenta:.2f}", "y la propina es de", f"{total_de_propina:.2f}", "por lo que el total a pagar es de", f"{total_a_pagar:.2f}", "y cada persona debe pagar", f"{pago_por_persona:.2f}")
```
