```python
suma = 0
while True:
    numero = float(input("Ingrese un número (0 para terminar): "))
    if numero == 0:
        break
    if numero > 0:
        suma += numero


print(f"la suma es: {suma: .4f}")
```

```python

horas = int(input("ingresa las horas acomuladas:"))

if horas <= 2:
    total = horas * 15.00
else:
    total = (2 * 15.00) + ((horas - 2)* 10.00)

if total > 200.0:
    total = 200.0 

print (f"total a pagar es: ${total: .2f}")
```

```python
numero = int(input("ingresa un numero entero: "))
num = numero 
invertido = 0

while num > 0:
    digito = num % 10
    invertido = invertido * 10 + digito 
    num //= 10

print (f"el numero invertido de {numero} es {invertido}")
```

```python

dividendo = int(input("ingrese el dividendo:"))
divisor = int(input("ingrese el divisor:"))

temp = dividendo 
cociente = 0

while temp >= divisor:
    temp -= divisor 
    cociente += 1

residuo = temp
print (f"{dividendo: .2f} ÷ {divisor: .2f} = {cociente: .2f} con residuo {residuo: .2f}")

```
