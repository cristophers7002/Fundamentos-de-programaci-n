## 1. Descripción del reto
Desarrollar una **aplicación en Python** que muestre una **tabla de Pitágoras
interactiva** para facilitar la multiplicación de dos factores ingresados por el
usuario. El programa debe almacenar la tabla en una **lista de listas (matriz)**,
imprimirla en pantalla de forma ordenada y permitir que el usuario capture dos
factores (renglón y columna) para obtener su producto consultando directamente la
información almacenada en la matriz.

```python
Tabla = int(input("de que tamañao la tabla: "))

matriz = []
for f in range(Tabla):
    fila_actual = []
    for c in range(Tabla):
        valor = (f + 1) * (c + 1)
        if f == c:
            valor = (f + 1) * (c + 1)
        fila_actual.append(valor)
    matriz.append(fila_actual)


for f in range (Tabla):
    for c in range (Tabla):
        print (f"{matriz[f][c]}", end ="\t")

    print ()


def multiplicar (matriz, fila, columna):
    return matriz[fila - 1] [columna - 1]

fila = int(input ("Ingrese un numero para multiplicar: "))
columna = int(input("ingrese un segundo numero;"))

resultado1 = multiplicar (matriz, fila, columna)

print(f"el resultado de {fila} por {columna} es: {resultado1}")
```

### Extra 3: Producto de una columna sin usar el operador *
**Enunciado:** Crea la matriz `[[1, 2, 3], [4, 5, 6], [7, 8, 9], [2, 4, 6]]` (4
filas y 3 columnas). Define una función `multiplicar_columna(matriz, columna)` que
**regrese valor** (con `return`) el producto de los elementos de una columna dada.
**Restricción:** dentro de la función está prohibido usar el operador `*`; el
producto debe calcularse con sumas repetidas usando un ciclo. Pide al usuario el
número de columna (0, 1, 2) y muestre el resultado

```python
matriz = [
    [1, 2, 3], 
    [4, 5, 6],
    [7, 8, 9], 
    [2, 4, 6]
]
for fila in matriz:
    for columna in fila:
        print(columna, end="   ")
    print()

def multiplicar(c, f):
    total = 0
    for i in range(f):
        total += c
    return total

def multiplicacion (matriz, columna):
    resultado = 1
    for fila in (matriz):
        resultado = multiplicar(resultado, fila[columna])
    return resultado

col = int(input("escoge un numero de la columna: 0, 1, 2"))

col1 = multiplicacion(matriz, 0)
col2 = multiplicacion(matriz, 1)
col3 = multiplicacion(matriz, 2)

match col:
    case 0:
        print(f"la multiplicacion de la columna 0 es:", col1)
    case 1:
        print(f"la multiplicacion de la columna 1 es:", col2)
    case 2:
        print(f"la multiplicacion de la columna 2 es:", col3)
```

### Extra 4: Tabla de Pitágoras de tamaño variable
**Enunciado:** Crea una función `generar_tabla(tamano)` que **regrese valor** (con
`return`) una lista de listas de `tamano x tamano` con los productos de la tabla de
Pitágoras (renglones y columnas del 1 al `tamano`). Dentro de la función de
generación está prohibido usar el operador `*`; usa sumas repetidas con ciclos.
Define una función `imprimir_tabla(tabla)` que **no regrese valor** e imprima la
tabla sin corchetes ni comas, con los números tabulados (`\t`). Pide al usuario un
tamaño entre 2 y 5; si ingresa un valor fuera del rango, muestra un mensaje de
error. Al final pide un renglón y una columna y muestra el producto consultando la
matriz, sin usar "*"

```python
def generar_tabla(fila):
    matriz = []
    for f in range(1, fila + 1):
        fila_actual = []
        for c in range(1, fila + 1):
            valor = 0
            for i in range(f):
                valor += c
            fila_actual.append(valor)
        matriz.append(fila_actual)
    return matriz

def imprimir_tabla(tabla):
    for fila in tabla:
        for valor in fila:
            print(f"{valor}", end="\t")
        print()
        
tam_tabla = int(input("tamaño de la tabla: "))

if tam_tabla < 2 or tam_tabla > 5:
    print("numeros del 2 al 5")
else:
    tabla = generar_tabla(tam_tabla)
    imprimir_tabla(tabla)

    renglon = int(input("Renglon: "))
    columna = int(input("Columna: "))

    
    resultado = tabla[renglon - 1][columna - 1]
    print(f"resultado de {renglon} por {columna} = {resultado}.")
```
