## 1. Descripción del reto
Crear una **aplicación interactiva en Python** que demuestre el uso integral de
**tuplas**, **diccionarios**, **manipulación de cadenas de texto (strings)** y
**manejo robusto de excepciones**, estructurada a través de un **menú principal
modular** controlado por un ciclo. El usuario debe poder seleccionar por número qué
sección desea ejecutar (Tuplas, Diccionarios, Excepciones, Strings o Finalizar) y
el programa debe delegar cada tarea a funciones modulares.
Este reto integra la creación y manipulación de **tuplas** (Tema 13), los
**diccionarios** con claves y valores (Tema 14), los bloques **try-except** con
excepciones específicas (Tema 15) y los **métodos de strings** con conteo de
palabras (Tema 16), además de consolidar las funciones con y sin retorno vistas en
semanas anteriores.

```python
numeros_tupla = (8, 60, 34, 0, 10)

contactos = {
    "Ofelia": "555-0192",
    "Eulalio": "555-0143", 
    "Carmen": "555-0178"
}

def sumar_elementos(tupla):
    return sum(tupla)

def buscar_telefono(dicc, nombre):
    return dicc.get(nombre, "No encontrado")

def contar_palabras(cadena):
    return len(cadena.split())

def tupla():
    print("tercer elemento:", numeros_tupla[2])
    try:
        num1 = float(input("primer numero extra: "))
        num2 = float(input("segundo numero extra: "))
        nueva_tupla = numeros_tupla + (num1, num2)
        
        lista_ordenada = list(nueva_tupla)
        lista_ordenada.sort()
        print("lista ordenada:", lista_ordenada)
        
        total_suma = sumar_elementos(nueva_tupla)
        print("suma total:", total_suma)
    except ValueError:
        print("error: ingrese numeros validos")

def telefonos():
    nuevo_nombre = input("nombre del nuevo contacto: ").strip().capitalize()
    nuevo_tel = input("telefono: ").strip()
    contactos[nuevo_nombre] = nuevo_tel
    
    print("contactos:")
    for nombre in contactos.keys():
        print(nombre)
        
    nombre_consulta = input("nombre a buscar: ").strip().capitalize()
    telefono = buscar_telefono(contactos, nombre_consulta)
    print("telefono:", telefono)

def division():
    try:
        num1 = float(input("primer numero: "))
        num2 = float(input("segundo numero: "))
        resultado = num1 / num2
        print(f"la division de {num1} entre {num2} es igual a: {resultado}")
    except ZeroDivisionError:
        print("Error: No se puede dividir entre cero.")

def palabra():
    mensaje = "tu novio te ama mucho"
    print("mensaje original:", mensaje)
    print("longitud:", len(mensaje))
    print("mayusculas:", mensaje.upper())
    
    mensaje_nuevo = mensaje.replace("tu novio", "Cris")
    print("mensaje nuevo:", mensaje_nuevo)
    
    total = contar_palabras(mensaje)
    print("numero de palabras:", total)

    print("1. Tupla")
    print("2. Telefonos")
    print("3. Division")
    print("4. Palabra")
    print("5. Salir")


opcion = ""
while opcion != "5":
    opcion = input("opcion: ").strip()
    
    match opcion:
        case "1":
            tupla()
        case "2":
            telefonos()
        case "3":
            division()
        case "4":
            palabra()
        case "5":
            print("bye")
        case _:
            print("opcion invalida")
```

## Sistema de calificaciones con tuplas
**Enunciado:** Crea una tupla `calificaciones = (7.5, 9.0, 8.0, 6.5, 10.0)`.
Imprime la tercera calificación, captura dos calificaciones nuevas con `input()` y
anéxalas en una nueva tupla. Convierte la nueva tupla en lista, ordénala **de mayor
a menor** y muestra la suma de todos los elementos usando una función que retorne
el valor.

```python
def calificaiones_suma (t):
    return sum(t)

calificaiones = (7.5, 9.0, 8.0, 6.5, 10.0)

print (f"la tercera calificaion es: {calificaiones[2]}")

nueva_cal = float(input("ingresa una calificaion exra:"))
nueva_cal2 = float(input(" ingresa la segunda calificacion:"))

calif = calificaiones + (nueva_cal, nueva_cal2)

total_calif = list(calif)
total_calif.sort(reverse=True)

print (f"lista ordenada: {total_calif}")
print (f"la primer calificaion agregada es: {nueva_cal}")
print (f"la segunda calificaion agregada es: {nueva_cal2}")

suma = calificaiones_suma(calif)

print (f"suma total de calificaiones: {suma:.2f}")
```

bloque `try-except` que capture `ValueError` (caracteres no numéricos o vacíos) con
un mensaje controlado y un bloque específico `except ZeroDivisionError` que muestre

## EXTRA CASTIGO 2
### Contador de votos con diccionario
Crea un diccionario `votos = {"Rojo": 0, "Azul": 0, "Verde": 0}`. Pide al usuario
el color de su voto cinco veces, suma uno al valor correspondiente en cada ocasión
y, al final, imprime con `items()` el color ganador junto con su total.

```python
candidatos = {
    "ROJO": 0, 
    "VERDE": 0, 
    "AZUL": 0
}

cantidad = int(input("cuantos votos van a ser?:"))
conteo = 1

while conteo <= cantidad:
    voto = input("ingresa el voto {conteo} (A, B o C):")
    if voto.upper () in candidatos:
        print (f"votos {conteo}")
        print (f"voto para {voto.upper()}")
        candidatos[voto.upper()] +=1
        conteo += 1
    else:
        print ("El voto no es valid, vota otra vez")

print ("\nresultados:")
for k, v in candidatos.items():
    print (f"Opcion {k}, votos: {v}")
```

## Extra castigo 5
### Minimenú modular integrador
Construye un menú con un ciclo `while` y opciones numéricas que permita ejecutar de
forma modular: (1) mostrar la tupla de números ordenada, (2) buscar un teléfono en
el diccionario `contactos`, (3) dividir dos números con manejo de
`ZeroDivisionError` y (4) analizar un mensaje de texto. La opción (5) debe
finalizar el programa.

```python
numeros_tupla = (45, 12, 89, 3, 27, 64)
contactos = {
    "Ana": "555-0192",
    "Carlos": "555-0143", 
    "Sofía": "555-0178"
    }

def ordenar_tupla():
    print("Tupla ordenada:", tuple(sorted(numeros_tupla)))

def busc_conta():
    nombre = input("Nombre: ").strip().capitalize()
    print("Teléfono:", contactos.get(nombre, "No encontrado"))

def division():
    try:
        num1 = float(input("primer numero: "))
        num2 = float(input("segundo numero: "))
        resultado = num1 / num2
        print(f"la division de {num1} entre {num2} es igual a: {resultado}")
    except ZeroDivisionError:
        print("Error: No se puede dividir entre cero.")

def texto():
    mensaje = input("Mensaje: ")
    print("Número de palabras:", len(mensaje.split()))

print("1. Mostrar tupla ordenada")
print("2. Buscar teléfono en contactos")
print("3. Dividir dos números")
print("4. Analizar mensaje de texto")
print("5. Salir")

opcion = ""
while opcion != "5":
    opcion = input("Opción:").strip()
    
    match opcion:
        case "1":
            ordenar_tupla()
        case "2":
            busc_conta()
        case "3":
            division()
        case "4":
            texto()
        case "5":
            print("bye")
        case _:
            print("seleccione una opcion bien.")
```
