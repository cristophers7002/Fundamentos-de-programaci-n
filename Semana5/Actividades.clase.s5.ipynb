```python
def suma_tupla(t):
    total = 0
    for e in t:
        total += e
        return total 

calorias = (800, 950, 700, 1300, 800, 900, 650)
resultado = suma_tupla(calorias)
print (f"metiste a tu cuerpo {resultado} calorias a tu cuerpo, gordo")
```

```python
try:
    dividiendo = int(input("dame un numero:"))
    divisor = int(input("entre que numero lo dividimos:"))
    resultado = dividiendo / divisor
except ZeroDivisionError:
    print ("error de la divicion entre 0")
else:
    print (f"el resultado es {resultado}")
finally:
    print ("fue un gusto hacer esa divicion, chido!")
```

```python
candidatos = {
    "A": 0, 
    "B": 0, 
    "C": 0
}

cantidad = int(input("cuantos votos van a ser?:"))
conteo = 1

while conteo <= cantidad:
    voto = input("ingresa el voto {conteo} (A, B o C):")
    if voto.upper () in candidatos:
        print (f"votos {conteo}")
        print (f"voto para {voto.upper()}")
        candidatos[voto.upper()] +=1
        conteo += 1
    else:
        print ("El voto no es valid, vota otra vez")

print ("\nresultados:")
for k, v in candidatos.items():
    print (f"Opcion {k}, votos: {v}")
```
