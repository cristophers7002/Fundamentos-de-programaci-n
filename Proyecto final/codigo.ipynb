```python
import time

Fecha = ()

def pantalla_carga():
    print("\n--- CARGANDO SISTEMA ---")
    for i in range(1, 6):
        print("Cargando... segundo", i)
        time.sleep(1)
    print("¡Listo! Carga completa.\n")


def pedir_fecha():
    global Fecha
    print("\n--- FECHA DE OPERACIÓN ---")
    dia = int(input("Ingresa el día: "))
    mes = int(input("Ingresa el mes: "))
    año = int(input("Ingresa el año: "))
    
    Fecha = (dia, mes, año)
    print("Fecha guardada correctamente:", Fecha)


def checar_inactividad():
    print("\nComprobando inactividad...")
    for i in range(3):
        time.sleep(0.5)
    
    resp = input("¿Sigues ahí? (si/no): ").lower()
    if resp == "no":
        return False
    return True


def ver_archivos():
    print("\n--- ARCHIVOS DISPONIBLES ---")
    print("1. menu_taqueria.txt")
    print("2. recetas.txt")
    print("3. proveedores.txt")
    print("4. historial_ventas.txt")
    
    opcion = input("¿Cuál archivo quieres abrir? (1-4): ")
  
    archivos = {
        "1": "menu_taqueria.txt",
        "2": "recetas.txt",
        "3": "proveedores.txt",
        "4": "historial_ventas.txt"
    }
    plantillas = {
        "menu_taqueria.txt": (
            "--- MENÚ TAQUERÍA CHAVA ---\n"
            "Tacos (Bistec, Longaniza, Costilla, Chuleta, Adobada, Campechanos): $25 c/u\n"
            "Refresco: $20\n"
            "Envase de cristal: $12"
        ),
        "recetas.txt": (
            "--- RECETAS DE SALSAS ---\n"
            "Salsa verde (tomate, chile serrano, aguacate, agua, sal)\n"
            "Salsa roja (chile de árbol, jitomate, agua y sal)"
        ),
        "proveedores.txt": (
            "--- DIRECTORIO DE PROVEEDORES ---\n"
            "Carnes: teléfono - 7129840880\n"
            "Lore: teléfono-7125568223\n"
            "Coca-cola: teléfono-5568906652"
        ),
        "historial_ventas.txt": "--- HISTORIAL DE VENTAS Y CORTES ---\n"
    }
    
    if opcion in archivos:
        nombre = archivos[opcion]
        try:
            archivo = open(nombre, "r", encoding="utf-8")
            contenido = archivo.read()
            print("\n--- CONTENIDO ---")
            print(contenido)
            archivo.close()
        except FileNotFoundError:
            print(f"\n[!] El archivo '{nombre}' no existía, creándolo automáticamente...")
            archivo = open(nombre, "w", encoding="utf-8")
            contenido_inicial = plantillas.get(nombre, "Archivo sin contenido.")
            archivo.write(contenido_inicial)
            archivo.close()
            
            print("\n--- CONTENIDO ---")
            print(contenido_inicial)
        except Exception as e:
            print("Ocurrió un error al abrir el archivo:", e)
    else:
        print("Opción no válida.")


def guardar_en_txt(texto):
    try:
        archivo = open("historial_ventas.txt", "a", encoding="utf-8")
        linea = f"Fecha: {Fecha} | Datos: {texto}\n"
        archivo.write(linea)
        archivo.close()
        print("Guardado en historial_ventas.txt exitosamente.")
    except:
        print("No se pudo guardar en el archivo.")


def mostrar_menu():
    matriz_menu = [
        ["1", "Registrar Pedido Nuevo"],
        ["2", "Ver Corte del Día"],
        ["3", "Gestión de Archivos TXT"],
        ["4", "Salir"]
    ]
    
    print("\n==================================")
    print("              MENÚ             ")
    print("==================================")
    for fila in matriz_menu:
        print(fila[0] + ". " + fila[1])
    print("==================================")


def main():
    print("=== BIENVENIDO AL SISTEMA DE LA TAQUERÍA ===")
    nickname = input("Ingresa tu nombre: ")
  
    bienvenida = "Hola " + nickname + ", bienvenido al sistema de caja."
    print("\n" + bienvenida)

    pantalla_carga()
    pedir_fecha()

    PRECIO_TACO = 25
    PRECIO_REFRESCO = 20
    PRECIO_CRISTAL = 12

    total_clientes = 0
    total_recaudado = 0
    total_tacos_vendidos = 0
    
    opcion = ""
    
    while opcion != "4":
        mostrar_menu()
        
        opcion = input("\nElige una opción (1-4) o presiona ENTER para revisar inactividad: ").strip()
        if opcion == "":
            activo = checar_inactividad()
            if not activo:
                print("Reiniciando el sistema...")
                nickname = input("Ingresa tu nickname otra vez: ")
                pedir_fecha()
            continue
        if opcion == "1":
            print("\n--- NUEVO PEDIDO ---")
            cliente = input("Nombre del cliente: ")
            tipos_tacos = ["Bistec", "Longaniza", "Costilla", "Chuleta", "Adobada", "Campechanos"]
            cantidades_tacos = {}
            cant_tacos = 0

            print("\n--- SELECCIÓN DE TACOS ---")
            for taco in tipos_tacos:
                respuesta = input(f"¿Quiere tacos de {taco}? (s/n): ").strip().lower()
                if respuesta == 's':
                    try:
                        cant = int(input(f"  --> ¿Cuántos tacos de {taco}?: ") or 0)
                        if cant < 0:
                            cant = 0
                    except ValueError:
                        print("  [!] Entrada no válida, se tomará como 0.")
                        cant = 0
                    cantidades_tacos[taco] = cant
                    cant_tacos += cant
                else:
                    cantidades_tacos[taco] = 0

            subtotal_tacos = cant_tacos * PRECIO_TACO
            
            print("\n¿Es para comer aquí o llevar?")
            print("1. Comer aquí\n2. Para llevar")
            donde = input("Opción: ")
            input("¿Con cilantro? (s/n): ")
            input("¿Con cebolla cruda? (s/n): ")
            input("¿Con cebolla cocida? (s/n): ")
            input("¿Qué salsa quiere? (roja/verde/sin salsa): ")
            subtotal_bebidas = 0
            subtotal_envases = 0
            
            quiere_bebida = input("\n¿Quiere refresco? (s/n): ").lower()
            if quiere_bebida == "s":
                try:
                    num_refrescos = int(input("¿Cuántos refrescos?: ") or 0)
                    num_envases = int(input("¿Cuántos envases de cristal se lleva?: ") or 0)
                    subtotal_bebidas = num_refrescos * PRECIO_REFRESCO
                    subtotal_envases = num_envases * PRECIO_CRISTAL
                except ValueError:
                    print("Error al ingresar cantidad de bebidas.")

            total_pedido = subtotal_tacos + subtotal_bebidas + subtotal_envases
            print("\n----------------------------------")
            print("CUENTA DE:", cliente)
            print("Detalle de tacos:")
            for taco, cant in cantidades_tacos.items():
                if cant > 0:
                    print(f" - {taco}: {cant}")
            
            print(f"Total tacos: {cant_tacos} (${subtotal_tacos})")
            if subtotal_bebidas > 0:
                print("Total refrescos:", f"${subtotal_bebidas}")
            if subtotal_envases > 0:
                print("Importe envases:", f"${subtotal_envases}")
            print("TOTAL A PAGAR: $", total_pedido)
            print("----------------------------------")
            total_clientes += 1
            total_recaudado += total_pedido
            total_tacos_vendidos += cant_tacos
            info = f"Cliente: {cliente} | Tacos: {cant_tacos} | Total: ${total_pedido}"
            guardar_en_txt(info)
        elif opcion == "2":
            print("\n----------------------------------")
            print("         CORTE DEL DÍA            ")
            print("----------------------------------")
            print("Fecha:", Fecha)
            print("Clientes atendidos:", total_clientes)
            print("Tacos vendidos:", total_tacos_vendidos)
            print("Dinero total acumulado: $", total_recaudado)
            print("----------------------------------")
            corte_info = f"CORTE -> Clientes: {total_clientes} | Total: ${total_recaudado}"
            guardar_en_txt(corte_info)
        elif opcion == "3":
            ver_archivos()
        elif opcion == "4":
            print("\nSaliendo del sistema... ¡Nos vemos,", nickname + "!")
        else:
            print("Opción incorrecta, intenta con un número del 1 al 4.")
        if opcion != "4":
            input("\nPresiona ENTER para regresar al menú...")

if __name__ == "__main__":
    main()
```
